```python
import random

with open('moby.txt') as file:
    book = file.read()
words = book.split()
uniqueWords = set(words)

following = {} #dictionary of a word and then the words that follow
for i in range(len(words) - 1): #loop through the entire book as a list
    word = words[i]
    nextWord = words[i + 1]
    if word not in following: #takes each word and what comes after it
        following[word] = [] # word

    following[word].append(nextWord) # after word


inputWord = input("enter a starting word: ")
inputNum = int(input("enter your sentence length: "))

sentence = [inputWord]
current = inputWord

while len(sentence) < inputNum:
    next = random.choice(following[current])
    sentence.append(next)
    current = next

print(" ".join(sentence), end=".")


```
